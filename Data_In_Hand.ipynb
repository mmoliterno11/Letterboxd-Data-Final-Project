# Data In Hand - Madeline Moliterno

The data for this project is sourced from my friend’s Letterboxd account, which tracks her movie-related activity. The goal is to analyze her preferences and make hypotheses about her favorite movie genres based on her interactions on the platform. When downloaded from Letterboxd, I used her account login. Since Letterboxd provides an export feature, the data is accurate and reliable, using her activity from the time she created her account until when I exported the data. 

When first downloaded, it gives a folder that contains files of CSV codes and other folders. Below shows my description of what these files are and how it was sectioned: 

CSV Files (In main folder):
These files contain structured data about her activity through:
- Comments
- Diary entries
- Profile information
- Ratings
- Reviews
- Watched movies
- Watchlist
  
Subfolders (In the Main Folder):
Three additional subfolders, each containing specific types of data:

- Deleted: Contains files for deleted items with CSV files of comments, diary entries, or reviews.
- Likes: Contains CSV files about her likes on films, lists, and reviews.
- Lists: Contains CSV files about her custom lists on Letterboxd.
- Orphaned: Contains files for diary entries, reviews, or comments that have lost their parent item. "Orphan" data typically happens when data/a film is deleted from Letterboxd.
